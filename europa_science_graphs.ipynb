```python
# Energy to melt a cylindrical hole through Europa ice
import numpy as np

# --- Parameters ---
radius     = 0.20        # m
depth      = 10_000      # m  (10 km)
T_initial  = 120        # K  (cold surface ice)
T_melt     = 273         # K
rho_ice    = 917         # kg/m³
c_avg      = 1400        # J/(kg·K)  avg specific heat 60→273 K
L_fusion   = 334_000     # J/kg  latent heat of fusion
g0         = 9.81        # m/s²

# --- Volume & mass ---
V = np.pi * radius**2 * depth
m = rho_ice * V

# --- Energy ---
dT     = T_melt - T_initial
E_heat = m * c_avg * dT          # sensible heat: warm ice to melt point
E_melt = m * L_fusion             # latent heat: melt the ice

E_total = E_heat + E_melt

print(f"Cylinder:        r={radius*100:.0f} cm, depth={depth/1e3:.0f} km")
print(f"Volume:          {V:.1f} m³")
print(f"Mass:            {m/1e6:.3f} Mt  ({m:.3e} kg)")
print()
print(f"ΔT:              {dT} K  ({T_initial} K → {T_melt} K)")
print(f"E_sensible:      {E_heat/1e9:.1f} GJ")
print(f"E_fusion:        {E_melt/1e9:.1f} GJ")
print(f"E_total:         {E_total/1e9:.1f} GJ  ({E_total/1e12:.3f} TJ)")
print()
print(f"= {E_total/3.6e9:.0f} MWh")
print(f"= {E_total/4.184e9:.0f} tonnes TNT")
print(f"@ 4 kW  → {E_total/4e3/3.156e7:.1f} years")
print(f"@ 40 kW → {E_total/4e4/86400:.0f} days")
```
