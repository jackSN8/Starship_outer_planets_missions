```python
import numpy as np
from astropy import units as u
from astropy.constants import hbar, k_B, a0, G, M_sun, M_earth, R_earth  # a0 = Bohr radius ~ 5.29e-11 m
```

```python
R_Mars=1.52*u.AU
r_parking=150*u.km

l_trans=np.sqrt(2*G*u.AU*M_sun*(1-(u.AU/(R_Mars+1*u.AU)).to(u.one)))
l_earth=np.sqrt(G*M_sun*u.AU)
print(l_earth)
print(l_trans)

dv_sun_frame=(l_trans-l_earth)/(u.AU)
print(dv_sun_frame.to(u.m/u.s))

v_eject=np.sqrt(dv_sun_frame**2+2*G*M_earth/(R_earth+r_parking)).to(u.m/u.s)
parking_orbit_speed=np.sqrt(G*M_earth/(r_parking+R_earth))

print(v_eject-parking_orbit_speed)
```
